###A importância das estruturas condicionais para programação

```python
x = 10

if x == 5:
  print("x é 5")
else:
  print("x não é 5")
```

```python
a = 8

resto = a % 2

if resto == 0:
  print("É par")
else:
  print("É impar")
```

```python
b = 9

if b > 10:
  print("B maior que 10")
elif b > 3:
  print("B maior que 3")
else:
  print("B nao é maior que 10")
```

```python
condicao = 3 > 6
condicao2 = 5 < 1


if condicao:
  # executo o codigo
  print("3 < 6")
  print("teste")
elif condicao2:
  print("5 > 1")
else:
  print("else final")
```

```python
num = int(input("Digite um numero: "))

if num > 0:
  print("Numero positivo")
elif num < 0:
  print("Numero negativo")
else:
  print("Numero é zero")
```

```python
saldo = 7835049

print("Menu")
print("1 - Ver saldo")
print("2 - Realizar depósito")
print("3 - Sair")

opcao = int(input("Digite uma opcao do menu"))

if opcao == 1:
  print(saldo)
elif opcao == 2:
  print("Deposito realizado")
elif opcao == 3:
  print("Saindo...")
else:
  print("Opcao inválida")
```

### O conceito de laços de repetição e o laço for em python

```python
for elemento in range(0, 11, 2):
  # bloco de execucao
  print(elemento)
```

```python
soma = 0

for numero in range(21):
  soma = soma + numero
  print("Soma parcial: ", soma)

print(soma)
```

```python
n = 8

fatorial = 1

for i in range(1, n+1):
  fatorial = fatorial * i

print(f"O fatorial de {n} é {fatorial}!")
```

### O laço while e Controle de fluxo em python

```python
contador = 5

while contador >= 0:
  print(contador)

  contador = contador - 1
```

```python
contador = 11

while contador > 0:
  if contador % 3 == 0:
    contador -= 1
    continue
  print(contador)
  contador -= 1
```

```python
i = 1

while i <= 20:
  if i % 2 == 0:
    i+= 1
    continue
  print("Numero impar:", i)
  i+=1
```

```python
soma = 0
numero = 1

while True:
  soma += numero
  if soma > 100:
    break
  numero +=1

print("Soma final ", soma)
```

### Boas Práticas sobre Laços de Repetição

```python
a = 1

for i in range(30):
  if i % 3 == 0:
    print(i)
    a += 1
  else:
    a += 2

print("A: ", a)
```

```python
contador = 8

# Repeticao enquanto o meu contador for maior do que 0
while contador > 0:
  print(contador)
  contador -= 1
```
