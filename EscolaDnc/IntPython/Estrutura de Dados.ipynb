### Manipulando dados em um dicionário

```python
dicionario_pet['idade'] = 5

print(dicionario_pet)
```

```python
dicionario_pet['animal'] = 'cachorro'

print(dicionario_pet)
```

```python
dicionario_pet['sexo'] = 'femea'

print(dicionario_pet)
```

```python
del dicionario_pet['cor']
print(dicionario_pet)
```

```python
tem_chave_cor = 'cor' in dicionario_pet
print(tem_chave_cor)
```

```python
tem_chave_sexo = 'sexo' in dicionario_pet
print(tem_chave_sexo)
```

```python
if 'cor' in dicionario_pet:
  print(dicionario_pet)
```

```python
dict_produto = {
    'nome': 'cadeira',
    'preco': 200,
    'estoque': 15
}

soma_lucro = 0

while dict_produto['estoque'] > 0:
  print(f"Posso vender, tenho {dict_produto['estoque']} unidades")

  dict_produto['estoque'] -= 1
  soma_lucro += dict_produto['preco']

print(dict_produto)
print(soma_lucro)
```
