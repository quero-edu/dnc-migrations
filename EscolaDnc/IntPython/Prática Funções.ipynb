### Passando argumentos para Funções

```python
def nome_funcao(argumento):
  # Corpo da função
  resultado = argumento * 2

  # Retorno da função
  return resultado
```

```python
nome_funcao(2)
```

```python
def saudar(nome):
  print(f"Olá {nome}, seja bem-vindo!")
```

```python
saudar('Bruno')
saudar('Maria')
saudar('João')
```

```python
def boas_vindas(nome, cidade):
  print(f"Seja bem-vindo, {nome} de {cidade}")
```

```python
boas_vindas("Bruno", "Sao Paulo")
```

```python
boas_vindas(cidade = "São Paulo", nome = "Bruno")
```

```python
def soma(a,b):
  return a + b
```

```python
soma(1, 9)
```

```python
def celsius_para_fahreinheit(celsius):
  return celsius * 9/5 + 32
```

```python
celsius_para_fahreinheit(0)
```

```python
def area_retangulo(largura, altura):
  area = largura * altura

  return area
```

```python
l = 10
a = 6

area_retangulo(l, a)
```

```python
area_retangulo(largura=l, altura=a)
```

```python
area_retangulo(altura=a, largura=l)
```

```python
def volume_cubo(largura, altura, comprimento):
  return largura * altura * comprimento

# Posicional
volume_pos = volume_cubo(43, 3, 5)

# Nomeada
volume_nom = volume_cubo(largura=43, comprimento=5, altura=3)
```

```python
print(volume_pos, volume_nom)
```

### Retornando valores de funções

```python
def par_impar(numero):
  if numero % 2 == 0:
    return 'par'
  else:
    return 'impar'
```

```python
par_impar(6)
```

```python
par_impar(9)
```

```python
def funcao_sem_retorno(x):
  print("Valor de x:", x)
```

```python
retorno_funcao = funcao_sem_retorno(6)
```

```python
print(retorno_funcao)
```

```python
def classifica_aluno(nota):
  if nota >= 7:
    return "Aprovado"
  elif 5 <= nota < 7:
    return "Recuperação"
  else:
    return "Reprovado"
```

```python
classifica_aluno(8)
```

```python
classifica_aluno(3)
```

```python
classifica_aluno(6)
```

###  Argumentos padrão, documentação e escopo

```python
def boas_vindas_padrao(nome, cidade='Sao Paulo'):
  """Envia uma mensagem de boas vindas

  Args:
    nome: uma string
    cidade: uma string (opcional, valor padrão é 'Sao Paulo')

  Retuns:
    Mensagem final personalizada

  """
  print(f"Seja bem-vindo, {nome} de {cidade}")
```

```python
print("mensagem1", "exemplo", sep='-', end=' ')
print("mensagem2")
```

```python
boas_vindas_padrao("Bruno", "Vitoria")
```

```python
boas_vindas_padrao('Maria')
```

```python
a = 10 # variavel global

def funcao():
  a = 5 # variavel local
  b = 8

  print(a, b)

funcao()
print(a)
print(b)
```

```python
contador = 0

def incrementar(c):
  contador = c
  contador += 1
  print("Valor de contador dentro da funcao", contador)


incrementar(contador)
print("Valor de contador fora da funcao", contador)
```

### Tratamento de erros

```python
try:
  print(x)

except:
  print("x não tem valor definido")
```

```python
def dividir(a, b):
  try:
    return a/b
  except:
    print("Erro: Divisão por 0")
```

```python
dividir(9, 0)
```

```python
def converte_texto_numero(texto):
  try:
    return int(texto)
  except:
    print(f"Erro de conversão. Não é possível converter {texto} para inteiro")
    return None
```

```python
converte_texto_numero("treze")
```
