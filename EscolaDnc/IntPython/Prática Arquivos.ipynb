### Aula 42 - Lendo arquivos TXT

```python
# Abertura do arquivo em modo de leitura r
arquivo = open('meu_arquivo.txt', 'r')

# Lendo o conteudo do arquivo
conteudo = arquivo.read()
print(conteudo)

# Fechamento do arquivo
arquivo.close()
```

```python
# Abertura do arquivo em modo de leitura r
arquivo = open('meu_arquivo.txt', 'r')

# Lendo linha a linha no arquivo - com for
for linha in arquivo:
  print("-", linha.strip())

# Fechamento do arquivo
arquivo.close()
```

```python
# Abertura do arquivo em modo de leitura r
arquivo = open('meu_arquivo.txt', 'r')

# Lendo linha a linha no arquivo - com while
linha = arquivo.readline()
while linha:
  print(linha.strip())
  linha = arquivo.readline()

# Fechamento do arquivo
arquivo.close()
```

```python
with open('meu_arquivo.txt', 'r') as arquivo:
  for linha in arquivo:
    print("-", linha.strip())
```

```python
# Quantas linhas tem no meu arquivo?

with open('mensagens.txt', 'r') as msg:
  linhas = msg.readlines()

  print("Numero de linhas: ", len(linhas))
```

### Aula 43 - Escrevendo em arquivos TXT

```python
with open('teste_escrita_w.txt', 'w') as arq_w:
  arq_w.write("Este é um teste de escrita\n")
  arq_w.write("Essa linha é a segunda do meu arquivo - teste de sobreescrita\n")
  arq_w.write("Toda vez que eu rodar esse codigo meu arquivo vai ser sobrescrito")
```

```python
with open('teste_escrita_a.txt', 'w') as arq_a:
  arq_a.write("Essa é a primeira linha escrita com a\n")
```

```python
with open('teste_escrita_a.txt', 'a') as arq_a:
  arq_a.write("Essa é a segunda linha do arquivo\n")
```

```python
lista_frutas = ['banana', 'ameixa', 'laranja', 'abacaxi']

with open('frutas.txt', 'a') as arq:
  for fruta in lista_frutas:
    arq.write(fruta + '\n')
```

### Aula 46 - Escrevendo arquivos JSON

```python
import json
```

```python
dados = {
    'nome': 'Carlos',
    'idade': 28,
    'cidade': "Rio de Janeiro"
}
```

```python
with open("dados.json", 'w') as arquivo:
  json.dump(dados, arquivo)
```

```python
dados2 = {
    'nome': 'Bruno',
    'idade': 30,
    'cidade': "Sao Paulo"
}
```

```python
lista_dados = [dados, dados2]
```

```python
lista_dados
```

```python
with open("dados.json", 'w') as arquivo:
  json.dump(lista_dados, arquivo)
```
